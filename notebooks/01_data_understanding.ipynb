```python
from sklearn.tree import DecisionTreeClassifier, plot_tree
import matplotlib.pyplot as plt
```

```python
features = [
    "AMT_INCOME_TOTAL",
    "AMT_CREDIT",
    "AMT_ANNUITY",
    "AMT_GOODS_PRICE"
]

X = df[features]
y = df["TARGET"]
```

```python
X_temp = X.dropna()
y_temp = y.loc[X_temp.index]
```

```python
tree = DecisionTreeClassifier(
    max_depth=3,
    min_samples_leaf=5000,
    random_state=42
)

tree.fit(X_temp, y_temp)
```

```python
plt.figure(figsize=(20, 10))

plot_tree(
    tree,
    feature_names=features,
    class_names=["No Default", "Default"],
    filled=True,
    rounded=True,
    proportion=True,
    precision=3
)

plt.show()
```

```python
financial_features = [
    "AMT_INCOME_TOTAL",
    "AMT_CREDIT",
    "AMT_ANNUITY",
    "AMT_GOODS_PRICE",
    "TARGET"
]

corr = df[financial_features].corr()

corr
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(9, 7))

im = ax.imshow(corr)

ax.set_xticks(range(len(corr.columns)))
ax.set_yticks(range(len(corr.columns)))

ax.set_xticklabels(corr.columns, rotation=45, ha="right")
ax.set_yticklabels(corr.columns)

for i in range(len(corr.columns)):
    for j in range(len(corr.columns)):
        ax.text(
            j,
            i,
            f"{corr.iloc[i, j]:.2f}",
            ha="center",
            va="center"
        )

plt.title("Correlation Matrix - Financial Features")
plt.colorbar(im)
plt.tight_layout()
plt.show()
```

# EDA Conclusions

## Dataset
- application_train.csv: 307,511 observations and 122 columns.
- SK_ID_CURR is the unique identifier.
- No duplicated rows or duplicated SK_ID_CURR values were identified.

## Target
- TARGET = 0: no payment difficulties.
- TARGET = 1: payment difficulties/default.
- Default rate is approximately 8.07%.
- The target is highly imbalanced.

## Missing Values
- 67 features contain missing values.
- 41 features have more than 50% missing values.
- Missing values will not be automatically removed because missingness itself may contain predictive information.
- OCCUPATION_TYPE missing observations showed a different default rate from non-missing observations.

## Numerical Features
- Higher age is associated with lower observed default rates.
- Financial variables such as AMT_CREDIT and AMT_ANNUITY show non-linear relationships with default.
- AMT_CREDIT and AMT_GOODS_PRICE are highly correlated (~0.99).
- Potential future engineered features include:
  - CREDIT / INCOME
  - ANNUITY / INCOME
  - CREDIT / ANNUITY
  - GOODS_PRICE / CREDIT

## Data Quality
- DAYS_EMPLOYED contains the anomalous value 365243 for 55,374 observations.
- This anomalous group has a different default rate and should not simply be discarded.
- A future transformation should replace the anomalous numerical value and preserve an anomaly indicator.

## External Scores
- EXT_SOURCE_1, EXT_SOURCE_2 and EXT_SOURCE_3 show the strongest linear correlations with TARGET.
- Higher EXT_SOURCE values are consistently associated with lower default rates.
- EXT_SOURCE_1 has substantial missingness but still appears potentially predictive.

## Categorical Features
- Cash loans have a higher observed default rate than revolving loans.
- Higher education levels are associated with lower observed default rates.
- Working clients show higher observed default rates than state servants and pensioners.
- Very small categories should not be interpreted based only on their raw default rates.
